```python
import tensorflow as tf
import numpy as np
from tensorflow import keras

print("TensorFlow Version:", tf.__version__)
```

```python
# Create a simple neural network

model = keras.Sequential([
    keras.layers.Input(shape=(1,)),
    keras.layers.Dense(8, activation='relu'),
    keras.layers.Dense(1)
])

model.summary()
```

```python
# Compile the model

model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.01),
    loss='mean_squared_error'
)
```

```python
# Training data
# Relationship approximately follows y = 2x + 5

x_train = np.array([0, 1, 2, 3, 4, 5, 6, 7], dtype=float)

y_train = np.array([
    5, 7, 9, 11, 13, 15, 17, 19
], dtype=float)

print("Input:", x_train)
print("Output:", y_train)
```

```python
# Train the neural network

history = model.fit(
    x_train,
    y_train,
    epochs=300,
    verbose=0
)

print("Training completed successfully.")
```

```python
# Test the trained model

test_values = np.array([8, 9, 10], dtype=float)

predictions = model.predict(test_values)

for value, prediction in zip(test_values, predictions):
    print(
        f"Input: {value:.0f}  ->  Predicted Output: {prediction[0]:.2f}"
    )
```

## Result

A simple feed-forward neural network was created using TensorFlow and Keras.

The network was trained using numerical input-output pairs that approximately follow the relationship:

y = 2x + 5

After training, new input values were supplied to the neural network. The model learned the relationship from the training data and generated predictions for unseen values.

This experiment demonstrates the basic working of an artificial neural network for regression.
