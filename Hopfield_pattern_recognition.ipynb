```python
import numpy as np

# Patterns for digits 0, 1 and 2
# 1 represents filled pixel and -1 represents empty pixel

digit_0 = np.array([
    1, 1, 1,
    1, -1, 1,
    1, 1, 1
])

digit_1 = np.array([
    -1, 1, -1,
    1, 1, -1,
    -1, 1, -1
])

digit_2 = np.array([
    1, 1, 1,
    -1, -1, 1,
    1, 1, 1
])

patterns = np.array([digit_0, digit_1, digit_2])

print("Stored Patterns:")
for i, pattern in enumerate(patterns):
    print(f"\nDigit {i}:")
    print(pattern.reshape(3, 3))
```

```python
# Create the Hopfield weight matrix

number_of_neurons = 9
weights = np.zeros((number_of_neurons, number_of_neurons))

for pattern in patterns:
    weights += np.outer(pattern, pattern)

# Neurons should not connect to themselves
np.fill_diagonal(weights, 0)

print("Weight Matrix:")
print(weights)
```

```python
def display_pattern(pattern):
    print(pattern.reshape(3, 3))


def hopfield_update(input_pattern, iterations=5):
    state = input_pattern.copy()

    for _ in range(iterations):
        for neuron in range(len(state)):
            activation = np.dot(weights[neuron], state)

            if activation > 0:
                state[neuron] = 1
            elif activation < 0:
                state[neuron] = -1

    return state
```

```python
# Create a noisy version of digit 1

noisy_input = np.array([
    -1, 1, 1,
    1, -1, -1,
    -1, 1, -1
])

print("Noisy Input:")
display_pattern(noisy_input)

recovered_pattern = hopfield_update(noisy_input)

print("\nRecovered Pattern:")
display_pattern(recovered_pattern)
```

```python
# Find the closest stored pattern

distances = []

for pattern in patterns:
    distance = np.sum(recovered_pattern != pattern)
    distances.append(distance)

recognized_digit = np.argmin(distances)

print("\nHamming Distances:", distances)
print("Recognized Digit:", recognized_digit)
```

## Result

The Hopfield Neural Network stores the patterns of digits 0, 1 and 2.

A noisy version of digit 1 is provided as input. The network repeatedly updates the neurons using the calculated weight matrix.

After convergence, the recovered pattern is compared with the stored patterns using Hamming distance.

The pattern having the smallest distance is selected as the recognized digit.

This demonstrates how a Hopfield Network can perform associative memory and pattern recognition.
